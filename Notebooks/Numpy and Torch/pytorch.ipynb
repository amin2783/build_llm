Numpy is faster than list or looping. Because ndarray uses contiguous block of memory, rather than random locations at memory and then pointing to them.

But that means NumPy arrays have a fixed size and must be homogeneous (all elements of same type)

```python
def python_forloop_list_approach(x, w):
    z = 0.
    for i in range(len(x)):
        z += x[i] * w[i]
    return z

a = [1., 2., 3.]
b = [4., 5., 6.]

large_a_list = list(range(10000))
large_b_list = list(range(10000))

%timeit python_forloop_list_approach(large_a_list, large_b_list)
```

Let's try using numpy's dot() method.

```python
import numpy as np

def numpy_dotproduct_approach(x, w):
    # same as np.dot(x, w)
    # same as x @ w
    return x.dot(w)


large_a_ndarray = np.arange(10000)
large_b_ndarray = np.arange(10000)

%timeit numpy_dotproduct_approach(large_a_ndarray, large_b_ndarray)
```

```python
ary = np.array([[1,2,3],
                [4,5,6]])

display(ary[0, -2]) # First row, second from last element

display(ary[-1, -1]) # Lower right

display(ary[1, 1])  # First row, second column

display(ary[:, 0]) # Entire first column

display(ary[:, :2]) # First two columns
```

### Ufuncs in NumPy
Used to vectorize operations.

```python
lst = [[1, 2, 3],
       [4, 5, 6]]  # 2d array

for row_idx, row_val in enumerate(lst):
    for col_idx, col_val in enumerate(row_val):
        lst[row_idx][col_idx] += 1

lst
```

```python
%%timeit

# the same can be achieved with list comprehensions

[[cell +1 for cell in row] for row in lst]
```

```python
ary = np.array([[1, 2, 3],
                [4, 5, 6]])
```

```python
%%timeit

# Easier still using NumPy's ufunc for element-wise scalar addition



ary2 = np.add(ary, 1)

ary
```

```python
np.add(ary, 1)
```

```python
# Same as np.add(ary, 1)

ary + 1
```

```python
np.power(ary, 2)
```

```python
# Same as np.power(ary, 2)

ary**2
```

To compute the sum or product of array element along a given axis, we can use `reduce`. By default, `reduce` applies an operation along the first axis (`axis=0`). In the case of a two-dimensional array, we can think of the first axis as the rows of a matrix.

```python
# Adding up elements along rows yields the column sums of that matrix

ary = np.array([[1, 2, 3],
                [4, 5, 6]])

np.add.reduce(ary, axis=0)
```
